```python
import numpy as np
import colorama
print(f"{colorama.Fore.RED}{colorama.Style.BRIGHT}===============> Welcome to Marks Analyzer <======================== {colorama.Style.RESET_ALL}\n")

print(f"{colorama.Fore.YELLOW}{colorama.Style.BRIGHT}Student Report{colorama.Style.RESET_ALL}\n")
names=np.array(["Syed Salman Ali","Ahmad","Ali","Sufyan","Khurram","Asif"])
subjects=np.array(["Engl","Math","Comp","Chem","Phys","Urdu"])

marks=np.array([
    [88,96,90,78,88,77],
    [66,88,38,69,88,61],
    [58,59,76,89,66,74],
    [55,87,62,54,88,98],
    [13,59,12,18,90,13],
    [41,76,56,98,67,58]
])

Total_marks=marks.sum(axis=1)
averages=marks.mean(axis=1).round(2)
Total_Marks=np.array(Total_marks)
Average=np.array(averages)
percentage=np.array((marks*100)/100.0)

def grade(G):
    if G>=80.0:
        return ("A")
    elif G>=70.0:
        return ("B")
    elif G>=60.0:
        return("C")
    elif G>=50.0:
        return("D")
    elif G>=40.0:
        return("E")
    else:
        return ("F")

        
for n in range(len(names)):
    print(f"\n{n+1}) {colorama.Fore.BLACK}{colorama.Style.BRIGHT}{names[n]}{colorama.Style.RESET_ALL}\n")
    print(f"{colorama.Fore.BLUE}{colorama.Style.BRIGHT}Subj :       Marks        Perc        Grade{colorama.Style.RESET_ALL}")
    for i in range(6):
        print(f"{colorama.Back.MAGENTA}{subjects[i]} :       {marks[n,i]}           {percentage[n,i]}%         {grade(percentage[n,i])}   {colorama.Style.RESET_ALL}")
    print(f"\nTotal: {colorama.Fore.MAGENTA}600{colorama.Style.RESET_ALL}")
    print(f"Obtained Marks: {colorama.Fore.MAGENTA}{Total_marks[n]}{colorama.Style.RESET_ALL}")
    print(f"Average: {colorama.Fore.MAGENTA}{averages[n]}{colorama.Style.RESET_ALL}")
    print(f"Grade : {colorama.Fore.MAGENTA}{grade(averages[n])}{colorama.Style.RESET_ALL}")

    
index=averages.argmax()
print(f"\n{colorama.Fore.YELLOW}{colorama.Style.BRIGHT}Student Analysis{colorama.Style.RESET_ALL}\n")
print(f"{colorama.Style.BRIGHT}Top student : {colorama.Fore.GREEN}{names[index]}{colorama.Style.RESET_ALL}")
print(f"{colorama.Style.BRIGHT}Average : {colorama.Fore.GREEN}{averages[index]}{colorama.Style.RESET_ALL}")
print(f"{colorama.Style.BRIGHT}Grade : {colorama.Fore.GREEN}{grade(averages[index])}{colorama.Style.RESET_ALL}")


index=averages.argmin()
print(f"\n{colorama.Style.BRIGHT}Lowest student : {colorama.Fore.GREEN}{names[index]}{colorama.Style.RESET_ALL}")
print(f"{colorama.Style.BRIGHT}Average : {colorama.Fore.GREEN}{averages[index]}{colorama.Style.RESET_ALL}")
print(f"{colorama.Style.BRIGHT}Grade : {colorama.Fore.GREEN}{grade(averages[index])}{colorama.Style.RESET_ALL}")


passed=np.array(averages[averages>40.0])
failed=np.array(averages[averages<40.0])
print(f"{colorama.Style.BRIGHT}Passed Student : {colorama.Fore.GREEN}{len(passed)}{colorama.Style.RESET_ALL}")
print(f"{colorama.Style.BRIGHT}Failed Student : {colorama.Fore.GREEN}{len(failed)}{colorama.Style.RESET_ALL}")

print(f"\n{colorama.Fore.YELLOW}{colorama.Style.BRIGHT}Subject Analysis Analysis{colorama.Style.RESET_ALL}")
m=np.sort(marks,axis=0)
for n in range(6):
    print(f"\n{colorama.Fore.GREEN}{colorama.Style.BRIGHT}{subjects[n]}{colorama.Style.RESET_ALL}")
    print(f"\n{colorama.Style.BRIGHT}Average : {colorama.Fore.GREEN}{marks.mean(axis=0)[n].round(2)}{colorama.Style.RESET_ALL}")
    print(f"{colorama.Style.BRIGHT}Highest Marks : {colorama.Fore.GREEN}{m[len(names)-1,n]}{colorama.Style.RESET_ALL}")
    print(f"{colorama.Style.BRIGHT}Highest scoring student : {colorama.Fore.GREEN}{names[marks.argmax(axis=0)[n]]}{colorama.Style.RESET_ALL}")
    print(f"{colorama.Style.BRIGHT}Lowest Marks : {colorama.Fore.GREEN}{m[0,n]}{colorama.Style.RESET_ALL}")
    print(f"{colorama.Style.BRIGHT}Lowest scoring student : {colorama.Fore.GREEN}{names[marks.argmin(axis=0)[n]]}{colorama.Style.RESET_ALL}")
```
